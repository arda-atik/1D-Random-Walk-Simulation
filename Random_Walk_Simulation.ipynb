```python
import random
import matplotlib.pyplot as plt

# 1. Define the initial variables (Başlangıç değişkenlerini tanımlıyoruz)
num_steps = 1000
current_position = 0
position_history = [current_position]  # We use a list to keep track of the path

# 2. The Simulation Loop (Simülasyon Döngüsü)
for i in range(num_steps):
    # Randomly choose a direction: -1 (backward) or 1 (forward)
    step_direction = random.choice([-1, 1])
    
    # Update the current position
    current_position = current_position + step_direction
    
    # Append the new position to our history list
    position_history.append(current_position)

# 3. Visualize the result (Sonucu görselleştiriyoruz)
plt.plot(position_history, color='blue', linewidth=1)
plt.title("1D Random Walk Simulation")
plt.xlabel("Number of Steps")
plt.ylabel("Position")
plt.show()
```
